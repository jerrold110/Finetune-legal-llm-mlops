```python
from mlrun import new_project
from mlrun import set_environment

from pathlib import Path

target_path = Path.cwd().parent #/ "mlrun-data" apparently mlrun automatically creates mlrun-data directory in the project root and uses it as the default artifact store, so we can just point to the project root as the artifact path
target_path = str(target_path.as_posix()) # convert windows path to unix path
abs_path = "file://" + target_path

p = set_environment("http://localhost:8080", artifact_path=abs_path)
```

# Create project

```python
# https://docs.mlrun.org/en/latest/api/mlrun.projects/index.html#mlrun.projects.new_project
# context is: project local directory path (The yaml file will be saved in this directory)

code_path = Path.cwd().parent
code_path = str(code_path.as_posix()) # convert windows path to unix path

project = new_project(
    name="finetune-legal-extractor", context=code_path, user_project=False, 
    init_git=False, description="A fine-tuned legal extractor LLM",
    overwrite=True
)
print(project.spec.get_code_path())
```

# delete project

```python
# This will delete the project and all its artifacts, functions, and workflows
from mlrun.common.schemas import DeletionStrategy
import mlrun

# Ensure your environment is connected to the MLRun API
mlrun.set_environment("http://localhost:8080")

# Fetch the database connection object
db = mlrun.get_run_db()

# Delete the project and all its internal records
db.delete_project(
    name="finetune-legal-extractor", 
    deletion_strategy=DeletionStrategy.cascade
)
```
